```python
# Barplot Python Seaborn -:

import seaborn as sns
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = sns.load_dataset('penguins')
df
```

```python
# sns.barplot(x = df.island , y = df.bill_length_mm)
# plt.show()

sns.barplot(x = 'island' , y = 'bill_length_mm', data = df,
            color = 'r')
plt.show()
```

```python
o = ['Dream', 'Biscoe', 'Torgersen']  
sns.barplot(x = 'island' , y = 'bill_length_mm', data = df, hue = 'sex', order = o,
           errorbar = ( 'ci' , 90), n_boot = 2)
plt.show()
```

```python
o = ['Dream', 'Biscoe', 'Torgersen'] 

sns.barplot(x = 'island' , y = 'bill_length_mm', data = df,
            hue = 'sex', order = o, 
            hue_order = ['Female','Male'])

plt.show()
```

```python
o = ['Dream', 'Biscoe', 'Torgersen']  
sns.barplot(x = 'bill_depth_mm' , y = 'bill_length_mm', data = df, 
            orient = 'h')
plt.show()
```

```python
o = ['Dream', 'Biscoe', 'Torgersen'] 

sns.barplot(x = 'island' , y = 'bill_length_mm', data = df, 
            hue = 'sex', order = o, 
            hue_order = ['Female','Male'],
            palette = 'icefire'
           )

plt.show()
```

```python
sns.set(style = 'darkgrid')
o = ['Dream', 'Biscoe', 'Torgersen']  
sns.barplot(x = 'island' , y = 'bill_length_mm', data = df,  hue = 'sex',
            saturation = 100,
            err_kws = {'color' : 'g'},
            capsize = 0.3,
            dodge = True,
            alpha = 0.7)


plt.show()
```
