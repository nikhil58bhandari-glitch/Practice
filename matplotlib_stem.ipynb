```python
import matplotlib.pyplot as plt 
x = [1,2,3,4,5,6]
y = [2,3,2,4,5,5]

plt.stem(x,y, linefmt = ':', markerfmt = 'ro',bottom = 1)
plt.show()
```

```python
import matplotlib.pyplot as plt 
x = [1,2,3,4,5,6]
y = [2,3,2,4,5,5]

plt.stem(x,y, basefmt = 'g', label = 'python', orientation = 'horizontal')

plt.legend()
plt.show()
```
