```python
# Strip plot -:
import pandas as pd 
import numpy as mp
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df =  sns.load_dataset('tips')
df 
```

```python
m = {'Male' : 'o', 'Female' : '>'}
sns.stripplot(x = "day", y = "total_bill", data = df, hue = "sex",
             marker = '>')
plt.show()
```

```python
sns.stripplot(x = "day", y = "total_bill", data = df, hue = "sex", 
              palette = 'rocket_r',
             linewidth = 1,
             edgecolor = 'r',
              # jitter = 1,
              size = 5,
              marker = '>',
              alpha = 0.5,
              
             )
plt.show()
```

```python
sns.stripplot(x = df["total_bill"])
plt.show()

# sns.stripplot(y = df["total_bill"])
# plt.show()
```

```python
# Box Plot -:

import pandas as pd 
import numpy as mp
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df =  sns.load_dataset('tips')
df 
```

```python
sns.set(style = 'whitegrid')
sns.boxplot(x = 'day', y = 'total_bill', data = df,
           color = 'g')
plt.show()
```

```python

sns.set(style = 'whitegrid')
sns.boxplot(x = 'day', y = 'total_bill', data = df,
           hue = 'sex',
           order = ['Fri', 'Sat','Thur', 'Sun'],
           showmeans = True,
           meanprops = {'marker' : '+', 'markeredgecolor' : 'r'},
           linewidth =  2,
           palette = 'plasma',
            orient = 'v',
            
           )
plt.show()
```

```python
sns.set(style = 'whitegrid')
sns.boxplot(x = 'total_bill',data = df)
plt.show()
```

```python
sns.set(style = 'whitegrid')
sns.boxplot(y = 'total_bill',data = df)
plt.show()
```
