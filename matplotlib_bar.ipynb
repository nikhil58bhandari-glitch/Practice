```python
import matplotlib.pyplot as plt

x = ['python','c','c++','java']
y = [85,70,60,82]
plt.xlabel('language',fontsize = 20)
plt.ylabel('no', fontsize = 15)
plt.title('wscube' ,fontsize = 20)
c = ['b','r','m','g']
plt.bar(x,y, width = 0.4, color = c, align = 'center', edgecolor = 'b', linewidth = 2, linestyle = ':', alpha =0.6,label = 'popularity')
plt.legend()
plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np 

x = ['python','c','c++','java']
y = [85,70,60,82]
z = [40,50,30,59]

width = 0.2

p = np.arange(len(x))
p1 = [j + width for j in p]

plt.xlabel('language',fontsize = 20)
plt.ylabel('no', fontsize = 15)
plt.title('wscube' ,fontsize = 20)

plt.bar(p,y, width , color = 'r', label = 'popularity' )
plt.bar(p1,z, width , color = 'y', label = 'popularity1')

plt.xticks(p+width/2, x,rotation = 20)
plt.legend()
plt.show()
```
