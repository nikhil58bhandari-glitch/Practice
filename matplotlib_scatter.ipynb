```python
import matplotlib.pyplot as plt 

x = [1,2,3,4,5,6,7]
y = [3,4,5,6,8,9,5]

plt.scatter(x,y,color = 'r')

plt.show()
```

```python
import matplotlib.pyplot as plt

day = [1,3,5,7,9,1,13,5]
no =  [4,6,7,2,5,8,9,2]

color = ['r','y','b','y','r','y','g','b']
size = [150,130,165,170,142,178,159,141]
plt.title('Scatter plot', fontsize = 15)
plt.xlabel('Day', fontsize = 15)
plt.ylabel('No.',fontsize = 15)

plt.scatter(day,no, c = color,s = size, alpha = 0.9, marker = '*', edgecolor = 'g',linewidth = 2)
plt.show()
```

```python
import matplotlib.pyplot as plt

day = [1,3,5,7,9,1,13,5]
no =  [4,6,7,2,5,8,9,2]

color = [10,40,30,46,34,64,75,87]
size = [150,130,165,170,142,178,159,141]

plt.title('Scatter plot', fontsize = 15)
plt.xlabel('Day', fontsize = 15)
plt.ylabel('No.',fontsize = 15)

plt.scatter(day,no, c = color,s = size, cmap = 'Accent' )
t = plt.colorbar()
t.set_label('colorbar', fontsize = 15)
plt.show()
```
