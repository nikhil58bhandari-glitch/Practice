```python
import matplotlib.pyplot as plt 

x = [10,20,30,40,50,60,70]

plt.boxplot(x, notch = False,patch_artist = True)
plt.show()
```

```python
import matplotlib.pyplot as plt 

x = [10,20,30,40,50,60,70]

plt.boxplot(x, widths = 0.2, tick_labels = ['pyhton'], showmeans = True)

plt.show()
```

```python
import matplotlib.pyplot as plt 

x = [10,20,30,40,50,60,70,120]

plt.boxplot(x, widths = 0.2, tick_labels = ['pyhton'], showmeans = True, whis = 2)

plt.show()
```

```python
import matplotlib.pyplot as plt 

x = [10,20,30,40,50,60,70,120]

plt.boxplot(x, widths = 0.2, tick_labels = ['pyhton'], showmeans = True, sym = 'g*', boxprops = dict(color = 'r'), capprops = dict(color = 'g'),
        whiskerprops = dict(color = 'm'), flierprops = dict(markeredgecolor = 'y')  )

plt.show()
```

```python
import matplotlib.pyplot as plt 

x = [10,20,30,40,50,60,70]
y = [10,20,40,50,70,80,90]


plt.boxplot(x, widths = 0.2, tick_labels = ['pyhton'], showmeans = True, sym = 'g*', boxprops = dict(color = 'r'), capprops = dict(color = 'g'),
        whiskerprops = dict(color = 'm'), flierprops = dict(markeredgecolor = 'y')  )

plt.boxplot(y, widths = 0.2, tick_labels = ['pyhton'], showmeans = True, sym = 'g*', boxprops = dict(color = 'r'), capprops = dict(color = 'g'),
        whiskerprops = dict(color = 'm'), flierprops = dict(markeredgecolor = 'y')  )

plt.show()
```

```python
x = [10,20,30,40,50,60,70]
y = [10,20,40,50,70,80,90]

z = [x, y]

plt.boxplot(z, tick_labels = ['python', 'java'], widths = 0.2, showmeans = True, sym = 'g*', boxprops = dict(color = 'r'), capprops = dict(color = 'g'),
        whiskerprops = dict(color = 'm'), flierprops = dict(markeredgecolor = 'y')  )



plt.show()
```
