```python
# Stack plot -: 
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6,7]
area0 = [3,4,5,2,6,2,1]
area1 = [2,4,1,5,6,2,4]
area2 = [1,2,4,2,3,5,6]

l = ['area0','area1','area2']

plt.stackplot(x,area0, area1, area2, labels = l, colors = ['r','g','m'], baseline = 'zero')
plt.title('python')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend(loc = 2)
plt.show()
```

```python
# Step plot -:
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,4,6,8,10,12]

plt.step(x,y)
plt.show()
```

```python
# Step plot -:
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,4,6,8,10,12]

plt.step(x,y, color = 'r',marker = 'o', ms = 10, mfc = 'g', label = 'pyhton')
plt.legend(loc = 4)
plt.title('python')
plt.xlabel('x- axis')
plt.ylabel('y-axis')
plt.grid()
plt.show()
```

```python
# Fill_Between plot-:
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5,6,7,8])
y = np.array([2,3,4,5,6,7,8,9])

plt.fill_between(x,y,color = 'm', where = (x >= 3) & (x <= 7),alpha = 0.5)
# plt.fill_between(x= [2,4], y1 = 2, y2 = 4)

plt.plot(x,y,color = 'r')
plt.title('python')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```
