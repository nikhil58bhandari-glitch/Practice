```python
# Violin Plot -: 

import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd 
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = df)
plt.show()
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = df,
              hue = 'time')
plt.show()
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = df,
              linewidth = 2,
              palette = 'Dark2',
              )
plt.show()
```

```python
sns.violinplot(x = 'time', y = 'total_bill', data = df,
               order = ["Dinner", "Lunch"], color = 'r',
               inner = 'quart'
              )
plt.show()
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = df, 
               hue = 'sex',
               split = True,
              density_norm = 'count')
plt.show()
```

```python
sns.violinplot(x = df["total_bill"])
plt.show()
```

```python
sns.violinplot(y = df["total_bill"])
plt.show()
```

```python
# Pair plot -: 

import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd 
```

```python
df = sns.load_dataset('tips').head(20)
df
```

```python
sns.pairplot(df)
plt.show()
```

```python
sns.pairplot(df, 
             hue = 'sex',
            # kind = 'kde',    # # scatter, kde, hist, reg
            # diag_kind = 'hist',
            markers = ['*', '<'],
            
            )  
plt.show()
```

```python
sns.pairplot(df, vars = ['total_bill','tip'],
             hue = 'sex',
            hue_order = ["Female","Male"],
             palette = "BuGn"
            )
plt.show()
```

```python
sns.pairplot(df,
             hue = 'sex',
            hue_order = ["Female","Male"],
             x_vars = ["total_bill", "tip"]
            )
plt.show()
```
