```python
# Line plot -:

import matplotlib.pyplot as plt 
import seaborn as sns 
import pandas as pd


var = [1,2,3,4,5,6,7]
var1 = [2,3,4,2,5,1,5]

x = pd.DataFrame({'var': var, 'var1' : var1})

# plt.plot(var ,var1)
# plt.show()

sns.lineplot(x = 'var', y = 'var1', data = x)

plt.show()
```
