```python
# Savefig matplotlib -:

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,3,4,2,6]

plt.plot(x,y,color = 'r')

# plt.savefig('line1', dpi = 2000 , facecolor = 'y')
plt.savefig('line1.pdf', dpi = 2000 , facecolor = 'y', transparent = False, bbox_inches = 'tight')

plt.show()
```

```python
# Axis matplotlib-:

import matplotlib.pyplot as plt 
x = [1,2,3,4,5]
y = [3,5,1,4,2]

plt.plot(x,y)

# plt.xticks(x, labels = ['python', 'c','c++','java','html'])
# plt.yticks(x)

# plt.xlim(0,10)
# plt.ylim(0,10)

plt.axis([0,10,0,7])
    
plt.show()
```

```python
# Text in matplotlib -: 

import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [3,2,4,1,5]

plt.plot(x,y)

plt.title('python' , fontsize = 15, color = 'r')
plt.xlabel('Days', fontsize = 15)
plt.ylabel('Growth', fontsize = 15)

# plt.text(2,4,'java', fontsize = 15, style = 'italic', bbox = {'facecolor': 'm'})

# plt.annotate('python', xy = (2,2),xytext = (3,3), arrowprops = dict(facecolor = 'black', shrink = 100))

plt.legend(['up'], loc = 9, facecolor = 'm')

plt.show()
```
