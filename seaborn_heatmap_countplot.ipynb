```python
# HeatMap Plot -:

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
var = np.linspace(1,10,20).reshape(4,5)
var
```

```python
y = {'fontsize' : 10, 'color' :'m'}
sns.heatmap(var, vmin = 0, vmax = 20, 
            cmap = 'gist_heat',
            annot = True,
            annot_kws = y,
            linewidth = 4,
            linecolor = 'y',
            cbar = True,
           
           )
plt.show()
```

```python
df = sns.load_dataset('anagrams')
x = df.drop(columns = 'attnr', axis = 1).head(10)
x
```

```python
sns.heatmap(x)

plt.show()
```

```python
y = {'fontsize' : 15, 'color' : 'm'}

sns.heatmap(x, vmin = 0, vmax = 20,
            cmap = 'icefire',
            annot = True,
            annot_kws = y,
            linewidth = 4,
           linecolor = 'r', 
           cbar = True, 
           xticklabels = True,
           yticklabels = True)

plt.show()
```

```python
y = {'fontsize' : 15, 'color' : 'm'}

v = sns.heatmap(x, vmin = 0, vmax = 20,
            cmap = 'icefire',
            annot = True,
            annot_kws = y,
            linewidth = 4,
           linecolor = 'r', 
           cbar = True, 
           xticklabels = True,
           yticklabels = True)

v.set(xlabel = 'python',
      ylabel = 'wscube')

sns.set(font_scale = 1)

plt.show()
```
