```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50]
y = ['c', 'c++', 'java', 'html','python']
c = ['r','b','g','y','m']

plt.pie(x, labels = y, colors = c, textprops = {'fontsize' : 15})


plt.show()
```

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50]
y = ['c', 'c++', 'java', 'html','python']
ex = [0.3,0.0,0.0,0.0,0.0]

plt.pie(x, labels = y, explode = ex, autopct = '%0.1f%%', shadow = True, radius = 1, labeldistance = 1.2)   


plt.show()
```

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50]
y = ['c', 'c++', 'java', 'html','python']
ex = [0.3,0.0,0.0,0.0,0.0]

plt.pie(x, labels = y, explode = ex, autopct = '%0.1f%%', startangle = 0, textprops = {'fontsize' : 15},
       counterclock = True, wedgeprops = {'linewidth' : 4, 'edgecolor' : 'm'} , rotatelabels = False )  

plt.title('wscube', fontsize = 15)
plt.legend(loc = 3)
plt.show()
```
